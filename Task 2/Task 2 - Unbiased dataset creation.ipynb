# **Deductive Logic in Language Models: Horizontal vs. Vertical Reasoning - Task 2 Unbiased dataset creation**

This notebook generates a bias-corrected version of the first 150,000 examples in the original *dataset.txt* file released with [J. Brinkmann, A. Sheshadri, V. Levoso, P. Swoboda, and Christian Bartelt, *"A Mechanistic Analysis of a Transformer Trained on a Symbolic Multi-Step Reasoning Task"*, arXiv:2402.11917, 2024](https://arxiv.org/abs/2402.11917). The bias is described in Appendix D of [D. Maltoni and M. Ferrara, *"Deductive Logic in Language Models: Horizontal vs. Vertical Reasoning"*, Machine Learning and Knowledge Extraction, 2026](https://doi.org/10.3390/make8070214).

Note: This step is optional because a bias-corrected version of the first 150,000 examples from the original dataset is already available in the repository as *dataset_150k_remapped_paper.txt*. However, this notebook provides a useful reference for understanding how the identified bias was corrected.

The following code imports the functions required to run the notebook.

```python
from dataset_utils import remap_line
```

The following variables specify the original dataset file to be processed and the output file in which the first 150,000 remapped examples will be saved.

```python
original_dataset_file_name = "dataset.txt"
output_dataset_file_name = "dataset_150k_remapped.txt"
```

Running the cell below creates a new file containing a bias-corrected version of the first 150,000 examples from the original dataset.

```python
n_states = 16
vocal_size = 16
n = 150000

with open(original_dataset_file_name, 'r', encoding='utf-8') as fin, open(output_dataset_file_name, 'w', encoding='utf-8') as fout:
    for i, line in enumerate(fin):
        if i >= n:
            break

        new_line = remap_line(line,vocal_size,n_states)
        fout.write(new_line)
```

The obtained file may differ slightly from the bias-corrected dataset used in the paper (available in the repository as *dataset_150k_remapped_paper.txt*) because the *remap_line* function includes a stochastic component. Nevertheless, the two files should be very similar and can be used interchangeably for training purposes.
