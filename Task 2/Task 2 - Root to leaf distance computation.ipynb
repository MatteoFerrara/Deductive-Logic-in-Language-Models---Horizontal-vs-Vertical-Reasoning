# **Deductive Logic in Language Models: Horizontal vs. Vertical Reasoning - Task 2 Root to leaf distance computation**

This notebook generates a file containing the root-to-leaf distances for each tree in the input dataset. The resulting file is required to reproduce the experiments described in Section 8 of [D. Maltoni and M. Ferrara, *"Deductive Logic in Language Models: Horizontal vs. Vertical Reasoning"*, Machine Learning and Knowledge Extraction, 2026](https://doi.org/10.3390/make8070214).

Note: This step is optional because the file containing the root-to-leaf distances used in the paper is already available in the repository as *dataset_150k_remapped_depths.txt*. However, this notebook provides a useful reference for understanding how these distances were computed.

The following code imports the functions required to run the notebook.

```python
from dataset_utils import compute_dataset_depths
```

The following variables specify the input dataset file to be processed and the output file in which the root-to-leaf distances will be saved.

```python
input_dataset_file_name = "dataset_150k_remapped_paper.txt"
output_dataset_depths_file_name = "dataset_150k_remapped_depths.txt"
```

Running the cell below creates a new file containing the root-to-leaf distances for each tree in the input dataset.

```python
depths = compute_dataset_depths(input_dataset_file_name)

with open(output_dataset_depths_file_name, 'w') as f:
    for item in depths:
        f.write(str(item) + '\n')
```
