# **Deductive Logic in Language Models: Horizontal vs. Vertical Reasoning - Task 2 Dataset without CoT creation**

This notebook creates a new version of the input dataset with the Chain-of-Thought (CoT) information removed. The resulting file is required to reproduce the experiments described in Section 7.2 of [D. Maltoni and M. Ferrara, *"Deductive Logic in Language Models: Horizontal vs. Vertical Reasoning"*, Machine Learning and Knowledge Extraction, 2026](https://doi.org/10.3390/make8070214).

Note: This step is optional because the CoT-free dataset file used in the paper is already available in the repository as *dataset_150k_remapped_without_CoT_paper.txt*. However, this notebook provides a useful reference for understanding how the CoT information was removed (see ).

The following code imports the classes and the functions required to run the notebook.

```python
from dataset_utils import BinaryTree, break_tree, check_double_root_in_tree
```

The following variables specify the input dataset file to be processed and the output file in which the dataset without CoT information will be saved.

```python
input_dataset_file_name = "dataset_150k_remapped_paper.txt"
output_dataset_file_name = "dataset_150k_remapped_without_CoT.txt"
```

Running the cell below creates the CoT-free dataset following the procedure described in Section 3.2 of [D. Maltoni and M. Ferrara, *"Deductive Logic in Language Models: Horizontal vs. Vertical Reasoning"*, Machine Learning and Knowledge Extraction, 2026](https://doi.org/10.3390/make8070214).

```python
no_two_roots_count = 0
none_count = 0
no_connectivity_count = 0

with open(input_dataset_file_name, 'r') as fin, open(output_dataset_file_name, 'w') as fou:
    for line in fin:
        line = line.strip()
        new_line = break_tree(line, verbose=False)
        if new_line is not None:
            tree_part, rest_part = new_line.split('|')
            if check_double_root_in_tree(tree_part):
                if new_line.endswith('Y'):
                    leaf = rest_part.split(':')[0]
                    root = rest_part.split(':')[1][:-1]
                    new_tree = BinaryTree(tree_part)
                    if new_tree.check_connectivity(root, leaf):
                        fou.write(new_line + "\n")
                    else:
                        no_connectivity_count += 1
                else:
                    fou.write(new_line + "\n")
            else:
                no_two_roots_count += 1
        else:
            none_count += 1

print(f"Number of lines that did not have two roots after break: {no_two_roots_count}")
print(f"Number of lines for which break_tree returned None: {none_count}")
print(f"Number of lines with Y but no connectivity: {no_connectivity_count}")
```

The obtained file may differ slightly from the CoT-free dataset used in the paper (available in the repository as *dataset_150k_remapped_without_CoT_paper.txt*) because the *break_tree* function includes a stochastic component. Nevertheless, the two files should be very similar and can be used interchangeably for training purposes.
