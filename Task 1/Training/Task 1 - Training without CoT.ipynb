# **Deductive Logic in Language Models: Horizontal vs. Vertical Reasoning - Task 1 Training without CoT**

This notebook reproduces the experiments conducted using the **nanoGPT** architecture without Chain-of-Thought (CoT) reasoning, as reported in Section 7.1 of [D. Maltoni and M. Ferrara, *"Deductive Logic in Language Models: Horizontal vs. Vertical Reasoning"*, Machine Learning and Knowledge Extraction, 2026](https://doi.org/10.3390/make8070214). It relies on the following Python scripts:
- **logic_data.py** - provides functions to create, shuffle, and split the datasets used in the experimentation.
- **model.py** - contains a modified version of the nanoGPT model originally implemented by [Andrej Karpathy](https://github.com/karpathy/nanoGPT).
- **NanoGPTTraining.py** - defines the *nanogpt_training* function allowing multiple training executions (*run_count*) with the same number of epochs (**epochs**) on the same dataset.
- **EnvUtilities.py** - contains the *setup_environment* function to configure the environment before starting the experiments.

The following code imports all necessary modules and functions required to run this notebook.

```python
import matplotlib.pyplot as plt

from NanoGPTTraining import nanogpt_training
```

The following parameters configure the dataset generation process and the nanoGPT architecture used in the experiment.

```python
# dataset configuration
dataset_config = {
    "literals": 6,
    "alphabet": 20,
    "dataset_size": 4096,
    "validation_perc": 0.25,
    "cot": False, 
}

# model configuration
model_config = {
    "n_layer": 6,
    "n_head": 8,
    "n_embd": 128,
    "dropout": 0.1,
    "bias": False,
    "use_mlp": True,
}
```

The following code cell reproduces the experiment used to generate Figures A1 and A2. To save the weights of the trained models, set the *out_folder_path* parameter to the directory where you want them stored. Running the full experiment (100 runs, each with 150 epochs) may take several hours. To reduce the execution time, set the *run_count* parameter to 1.

```python
run_count=1
epochs=500
mb_size=128

run_train_seq_acc,run_val_seq_acc=nanogpt_training(run_count,epochs,mb_size,dataset_config = dataset_config, model_config = model_config)
```

Running the cell below reproduces the plot showing the model accuracy obtained without CoT information. The model achieves near-perfect accuracy on the training set but fails to generalise to the validation set.

```python
plt.plot(run_train_seq_acc[0]*100,label='Train')
plt.plot(run_val_seq_acc[0]*100,label='Val')
plt.xlabel('Epochs')
plt.ylabel('Sequence Accuracy (%)')
plt.title(f'The model accuracy without CoT')
plt.legend()

plt.show()
```
